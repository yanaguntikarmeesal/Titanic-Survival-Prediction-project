# **Titanic Survival Prediction project**

# Step 1: Import Libraries and Load the DatasetLibraries:
* We use pandas for data manipulation and handling.   
* Loading: We load the Titanic dataset from a CSV file

```python
# Import necessary libraries
import pandas as pd
import numpy as np
```

```python
# Load the Titanic dataset
df = pd.read_csv("titanic.csv")
```

```python
# Display the first few rows of the dataset to understand its structure[cite: 1]
print("Initial Dataset Preview:")
display(df.head())
```

# Step 2: Exploratory Data Analysis (EDA)
* Check the data types, non-null counts, and missing values for each column.

```python
# Check data types and non-null counts
df.info()
```

```python
# Check the total number of missing values in each column
print("\nMissing Values Count:")
print(df.isnull().sum())
```

# Step 3: Handle Missing Values
* Age: Missing values are filled with the mean age of the dataset

* Embarked: Missing values are filled with the most frequent value

```python
# Fill missing 'Age' values with the mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
# Fill missing 'Embarked' values with the mode (most common port)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
# Verify that all missing values for Age and Embarked are resolved
print("Remaining missing values in Age:", df['Age'].isnull().sum())
print("Remaining missing values in Embarked:", df['Embarked'].isnull().sum())
```

# Step 4: Drop Unnecessary Columns
Columns like PassengerId, Name, Ticket, and Cabin do not directly contribute useful numerical patterns for this baseline prediction model and are dropped.

```python
# Drop columns that are irrelevant or have too many missing values
df = df.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'])
```

```python
# View the updated dataset structure
df.head(2)
```

# Step 5: Encode Categorical Variables
* Machine learning models require numerical inputs. We convert text-based categories (Sex and Embarked) into numerical values.

```python
# Map 'Sex': male to 1, female to 0
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})
```

```python
# Map 'Embarked': 'S' to 1, 'C' to 2, 'Q' to 3
df['Embarked'] = df['Embarked'].map({'S': 1, 'C': 2, 'Q': 3})
```

```python
# View the fully numeric dataset
df.head(2)
```

# Step 6: Define Independent and Dependent Variables

* Independent variables
($x$): Features used to make predictions (Pclass, Sex, Age, SibSp, Parch, Fare, Embarked).   

* Dependent variable ($y$): The target label we want to predict (Survived).

```python
# Split the dataset into features (x) and target (y)
x = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
# Display feature and target previews
print("Features Preview:")
display(x.head(2))
print("Target Preview:")
display(y.head(2))
```

# Step 7: Train-Test Split
* Split the data so that 75% is used for training the model and 25% is reserved for testing its performance.

```python
from sklearn.model_selection import train_test_split

# Split data into training and testing sets (75% train, 25% test)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)

# Print the shapes of the split datasets
print("The x train shape is:", x_train.shape)
print("The x test shape is :", x_test.shape)
print("The y train shape is:", y_train.shape)
print("The y test shape is :", y_test.shape)
```

# Step 8: Train the Logistic Regression Model
* Initialize and fit a Logistic Regression model using the training data

```python
import warnings
warnings.filterwarnings('ignore')

from sklearn.linear_model import LogisticRegression

# Initialize the Logistic Regression model
model = LogisticRegression()

# Train the model using the training data
model.fit(x_train, y_train)

# Make predictions on the test set
y_pred = model.predict(x_test)
print("First 10 predictions:", y_pred[:10])
```

# Step 9: Evaluate Model Performance
* Evaluate how well the model predicts survival using accuracy score and a confusion matrix

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Calculate accuracy score[cite: 8]
accuracy = accuracy_score(y_test, y_pred)
print("Model Accuracy: {:.2f}%".format(accuracy * 100))

# Generate confusion matrix[cite: 8]
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

# Step 10: Save the Trained Model
* Use pickle to serialize and save the trained model into a file so it can be deployed later (e.g., in a Streamlit application).

```python
import pickle

# Save the trained model to disk
filename = 'titanic.pkl'
pickle.dump(model, open(filename, 'wb'))
print(f"Model successfully saved as {filename}")
```
